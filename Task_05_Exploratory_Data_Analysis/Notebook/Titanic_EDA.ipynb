```python
# Calculate Q1 and Q3
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
fare_outliers = df[
    (df['Fare'] < lower_bound) |
    (df['Fare'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(fare_outliers))
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Fill missing Embarked values with the mode
df_clean['Embarked'] = df_clean['Embarked'].fillna(
    df_clean['Embarked'].mode()[0]
)

# Verify missing values
print("Missing Embarked values:", df_clean['Embarked'].isnull().sum())
```

```python
# Median age by passenger class and gender
age_medians = df_clean.groupby(
    ['Pclass', 'Sex'],
    observed=True
)['Age'].median()

print(age_medians)
```

```python
# Fill missing Age using Pclass and Sex group medians
df_clean['Age'] = df_clean.groupby(
    ['Pclass', 'Sex']
)['Age'].transform(
    lambda x: x.fillna(x.median())
)

# Verify missing values
print("Missing Age values:", df_clean['Age'].isnull().sum())

# Check updated dataset
print("\nUpdated Age Summary:")
print(df_clean['Age'].describe())
```

```python
# Create a cabin availability feature
df_clean['Cabin_Known'] = df_clean['Cabin'].notnull().astype(int)

# Check distribution
print(df_clean['Cabin_Known'].value_counts())

print("\nCabin Availability Percentage:")
print(df_clean['Cabin_Known'].value_counts(normalize=True) * 100)
```

```python
# Survival rate based on cabin information availability
cabin_survival = pd.crosstab(
    df_clean['Cabin_Known'],
    df_clean['Survived'],
    normalize='index'
) * 100

print("Survival Percentage by Cabin Availability:")
print(cabin_survival)

# Visualization
cabin_survival.plot(
    kind='bar',
    stacked=True,
    figsize=(8, 5),
    color=['salmon', 'seagreen']
)

plt.title('Survival Rate by Cabin Information Availability')
plt.xlabel('Cabin Information (0 = Unknown, 1 = Available)')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(['Did Not Survive', 'Survived'])

plt.savefig(
    r"C:\Users\manis\Downloads\Elevate Intern\Task-5\Survival_by_Cabin.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
# Check missing values after cleaning
print("Missing Values After Cleaning:")
print(df_clean.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df_clean.duplicated().sum())

# Check final dataset shape
print("\nFinal Dataset Shape:", df_clean.shape)

# Display first five rows
df_clean.head()
```

```python
# Recreate AgeGroup using cleaned Age values
df_clean['AgeGroup'] = pd.cut(
    df_clean['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior']
)

# Verify missing values
print("Missing AgeGroup values:", df_clean['AgeGroup'].isnull().sum())

# Final missing-value summary
print("\nMissing Values:")
print(df_clean.isnull().sum())
```

```python
# Export cleaned dataset
df_clean.to_csv(
    r"C:\Users\manis\Downloads\Elevate Intern\Task-5\titanic_cleaned.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```

```python
# Calculate family size
df_clean['FamilySize'] = (
    df_clean['SibSp'] + df_clean['Parch'] + 1
)

# Categorize passengers
df_clean['FamilyType'] = df_clean['FamilySize'].apply(
    lambda x: 'Alone' if x == 1
    else 'Small Family' if x <= 4
    else 'Large Family'
)

# Survival rate by family type
family_survival = df_clean.groupby(
    'FamilyType',
    observed=True
)['Survived'].mean() * 100

print("Survival Rate by Family Type:")
print(family_survival)
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=family_survival.index,
    y=family_survival.values,
    hue=family_survival.index,
    palette='Set2',
    legend=False
)

plt.title('Survival Rate by Family Type')
plt.xlabel('Family Type')
plt.ylabel('Survival Rate (%)')
plt.ylim(0, 100)

plt.savefig(
    r"C:\Users\manis\Downloads\Elevate Intern\Task-5\Survival_by_FamilyType.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

## Final EDA Findings – Titanic Dataset

### 1. Dataset Overview
- Total passengers analysed: 891
- Total features in the cleaned dataset: 16
- Overall survival rate: 38.38%
- Overall non-survival rate: 61.62%

### 2. Key Findings
- **Gender:** Female passengers had a survival rate of 74.20%, compared with 18.89% for males.
- **Passenger Class:** First-class passengers had the highest survival rate (62.96%), while third-class passengers had the lowest (24.24%).
- **Age:** Children had a survival rate of 57.97%, while senior passengers had a survival rate of 22.73%.
- **Fare:** The fare distribution was highly right-skewed, with 116 potential outliers identified using the IQR method.
- **Cabin Information:** Passengers with recorded cabin information had a survival rate of 66.67%, compared with 29.99% for those without recorded cabin information.
- **Family Size:** Small-family passengers had the highest survival rate (57.88%), while large-family passengers had the lowest (16.13%).

### 3. Data Cleaning
- Filled missing Age values using median imputation grouped by passenger class and gender.
- Filled missing Embarked values using the mode.
- Retained missing Cabin values because approximately 77.10% were unavailable.
- Created Cabin_Known, AgeGroup, FamilySize and FamilyType features.
- No duplicate records were identified.

### 4. Conclusion
Exploratory Data Analysis revealed that gender, passenger class, age, fare and family size were associated with passenger survival outcomes. Female passengers, first-class passengers and passengers travelling with small families showed comparatively higher survival rates.

The analysis demonstrates how statistical summaries, visualizations and feature engineering can help identify meaningful patterns in real-world datasets.
